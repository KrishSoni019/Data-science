## for Loop

### Print list Though forLoop

```python
lst = ["Krish","Soni",20,"MIT WPU","MCA",1272250632]
```

```python
for i in lst:
    print(i)
```

### Basic Loop use

```python
#Print some Numbers.
for i in range(10):
    print(i)
```

```python
for i in range(5,10):
    print(i)
```

## While Loop

```python
r = 0
```

```python
while r<15:
    print(r)
    r+=1
```
