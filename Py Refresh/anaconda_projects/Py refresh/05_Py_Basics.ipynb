# Match Case

```python
a = 1
```

```python
match a:
    case 5:
        print("its 5.")
    case 9:
        print("its 9.")
    case 10:
        print("its 10.")
    case _:
        print("No match found.")
print("End of Program.")
```

## Formating number in Fstring

```python
pi = 3.14159
```

```python
print(f"Value of Pi is : {pi:.2f}")
```

## Alignment in Fstring

```python
print(f"{'Hello Krish.':<20}")
```

```python
print(f"{'Hello Krish.':>20}")
```

```python
print(f"{'Hello Krish.':^20}")
```
